# Código de Verificação
1.   Leitura dos dados;
2.   Execução das verificações;
2.   Resultado final impresso.

Dados hipotéticos baseados na tabela de telemetria

```python
temp_interna = float(input("Temperatura interna: "))
temp_externa = float(input("Temperatura externa: "))
integridade = int(input("Integridade estrutural (0 ou 1): "))
energia = float(input("Nível de energia (%): "))
pressao = float(input("Pressão dos tanques: "))
modulos = int(input("Status dos módulos (0 ou 1): "))

 # Verificações
pronto = True

if not (15 <= temp_interna <= 25):
    pronto = False
if not (20 <= temp_externa <= 320):
    pronto = False
if integridade != 1:
    pronto = False
if energia < 80:
    pronto = False
if not (400 <= pressao <= 600):
    pronto = False
if modulos != 1:
    pronto = False

# Resultado
if pronto:
    print("PRONTO PARA DECOLAR")
else:
    print("DECOLAGEM ABORTADA")

```

# Código de Análise Energética


*   Capacidade total (kwh);
*   Carga atual (%);
*   Consumo estimado na decolagem;
*   Perdas energéticas.







```python
capacidade = 120000
carga = 95
consumo = 20000
perdas = 100

energia_disponivel = capacidade * (carga / 100)
energia_real = energia_disponivel - perdas
autonomia = energia_real / consumo

print(f'Energia disponível: {energia_disponivel}')
print(f'Energia real: {energia_real}')
print(f'Autonomia estimada: {autonomia:.2f}')
```
